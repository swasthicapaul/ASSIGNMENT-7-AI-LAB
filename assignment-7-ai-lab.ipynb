Write a python programming to implement Adverserial Search (Minimax)

```python
def minimax(node, depth, maximizing_player, tree):

    # If the node is a leaf node, return its value
    if node not in tree:
        return node

    if maximizing_player:
        best_value = float('-inf')

        for child in tree[node]:
            value = minimax(child, depth + 1, False, tree)
            best_value = max(best_value, value)

        return best_value

    else:
        best_value = float('inf')

        for child in tree[node]:
            value = minimax(child, depth + 1, True, tree)
            best_value = min(best_value, value)

        return best_value



tree = {
    'A': ['B', 'C'],
    'B': [3, 5],
    'C': [2, 9]
}

# Start Minimax from root node A
best_value = minimax('A', 0, True, tree)

print("Best value for MAX:", best_value)
```
